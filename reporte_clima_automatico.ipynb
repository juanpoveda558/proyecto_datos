```python
#CODIGO COMPLETO INTEGRADO

import requests
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import time


# 1. Configuración - ¡ASEGÚRATE DE QUE TU API KEY ESTÉ AQUÍ!
API_KEY = " " 
CIUDADES = ["Bogota", "Medellin", "Cali"]
reporte = []

print("Iniciando consulta a la API...")

# 2. Obtención de datos
for ciudad in CIUDADES:
    url = f"http://api.openweathermap.org/data/2.5/weather?q={ciudad},CO&appid={API_KEY}&units=metric&lang=es"
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            data = response.json()
            reporte.append({
                "Ciudad": ciudad,
                "Temperatura (°C)": data['main']['temp'],
                "Humedad (%)": data['main']['humidity'],
                "Presion (hPa)": data['main']['pressure'],
                "Viento (m/s)": data['wind']['speed']
            })
            print(f"✅ Datos de {ciudad} obtenidos correctamente.")
        else:
            print(f"⚠️ Error en {ciudad}: Código {response.status_code} (Revisa tu API KEY)")
    except Exception as e:
        print(f"❌ Error de conexión en {ciudad}: {e}")

# 3. Verificación y Gráfica
if len(reporte) > 0:
    df = pd.DataFrame(reporte)
    
    # Guardar CSV en la ruta de Linux
    ruta_csv = os.path.expanduser("~/proyecto_datos/reporte_clima_final.csv")
    df.to_csv(ruta_csv, index=False)
    
    # Dashboard
    sns.set_theme(style="whitegrid")
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle('Dashboard Climático Colombia', fontsize=16)

    # Usamos colores directos para evitar problemas con paletas antiguas
    sns.barplot(ax=axes[0], x="Ciudad", y="Humedad (%)", data=df, color="skyblue")
    axes[0].set_title("Humedad")

    sns.barplot(ax=axes[1], x="Ciudad", y="Presion (hPa)", data=df, color="salmon")
    axes[1].set_title("Presión Atmosférica")

    sns.barplot(ax=axes[2], x="Ciudad", y="Viento (m/s)", data=df, color="lightgreen")
    axes[2].set_title("Velocidad del Viento")

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
    print(f"\nProceso exitoso. Archivo guardado en: {ruta_csv}")
    display(df)
else:
    print("\n🚨 ERROR CRÍTICO: No se obtuvo ningún dato. La gráfica no se puede generar.")
    print("Por favor, verifica que tu API_KEY sea válida y tengas conexión a internet.")
```
